```python
import pandas as pd 
import pickle
from sklearn.linear_model import LogisticRegression

# load 
file_path= "X_train_vec.pkl"
try:
    with open(file_path, 'rb') as file:
        X_train_vec= pickle.load(file)
        print(X_train_vec)
except FileNotFoundError :
    print(f"Error: The file '{file_path}' was not found.")

y_train=pd.read_csv("y_train.csv")
print(y_train)

#make sure x_train & Y_train are equal
X_train_vec.shape[0]==len(y_train)

#Apply One-Hot Encoding to convert y-train to numeric
from sklearn.preprocessing import LabelEncoder
le = LabelEncoder()
y_train_encoded = le.fit_transform(y_train.values.ravel())
print(y_train_encoded)
#train the model
model= LogisticRegression(max_iter= 1000)
model.fit(X_train_vec, y_train_encoded)
model.predict(X_train_vec)
score= model.score(X_train_vec, y_train_encoded)
print(score)
with open('expense_model.pkl', 'wb') as saved_file:
    pickle.dump(model, saved_file)



```
